# 4. Unit of Analysis

The unit of analysis is **one content page**.

Each row in the dataset represents a single webpage along with its historical search and engagement signals, such as impressions, clicks, CTR, trend direction, and search volume.

The model will generate one refresh priority score for each page.

```python
df["refresh_priority"] = (
    (df["trend_direction"] == "down") &
    (df["impressions_90d"] > df["impressions_90d"].median())
).astype(int)

df[["trend_direction", "impressions_90d", "refresh_priority"]].head(10)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*# 5. Why ML Beats a Fixed Rule

A simple rule such as "refresh every page with declining traffic" ignores many important factors.

Some declining pages receive very little traffic, while others still have high impressions, search demand, or user engagement.

Machine learning can combine multiple signals simultaneously and learn which combinations best identify valuable refresh opportunities. This produces a more useful ranked list than a single threshold-based rule.
